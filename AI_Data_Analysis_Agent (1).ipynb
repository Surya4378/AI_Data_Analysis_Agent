```python
from google import genai
from google.colab import userdata

GEMINI_API_KEY = userdata.get("GEMINI_API_KEY")

client = genai.Client(api_key=GEMINI_API_KEY)

print("Gemini connected successfully ✅")
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("snehaanbhawal/resume-dataset")

print("Path to dataset files:", path)
```

```python
import os

print("Dataset path:")
print(path)

print("\nFiles inside the dataset:")
print(os.listdir(path))
```

```python
print("DATA folder:")
print(os.listdir(os.path.join(path, "data")))

print("\nRESUME folder:")
print(os.listdir(os.path.join(path, "Resume")))
```

```python
import pandas as pd
import os

csv_path = os.path.join(path, "Resume", "Resume.csv")

df = pd.read_csv(csv_path, encoding="latin1")

print("Dataset loaded successfully ✅")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.head()
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
print(df.info())
```

```python
print("\nFirst 3 rows:")
display(df.head(3))
```

```python
data_summary = f"""
Dataset shape:
Rows: {df.shape[0]}
Columns: {df.shape[1]}

Column names:
{df.columns.tolist()}

Data types:
{df.dtypes}

First 5 rows:
{df.head().to_string()}
"""

print(data_summary)
```

```python
prompt = f"""
You are an AI Data Analysis Assistant.

Analyze the following dataset information.

{data_summary}

Explain:
1. What this dataset contains
2. What each column represents
3. What kind of analysis can be performed on this dataset

Keep the explanation simple and clear.
"""

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=prompt
)

print(response.text)
```

```python
question = input("Ask a question about the dataset: ")

prompt = f"""
You are an AI Data Analysis Assistant.

Here is information about the dataset:

{data_summary}

User question:
{question}

Answer the user's question clearly and simply.
"""

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents=prompt
)

print("\n🤖 AI Answer:")
print(response.text)
```
